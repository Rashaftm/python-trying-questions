```python
# number separated with commas stored in tuple 
number = input("Enter the numbers separated with commas: ") 
print("The numbers are: ", number) 
num_t = tuple(map(int, number.split(','))) 
average = sum(num_t) / len(num_t) 
print("The average is: ", average)
```
